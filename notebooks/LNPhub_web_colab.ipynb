## Optional: add an ngrok token

A free ngrok token makes the tunnel more reliable. Add it below, or leave it empty to try an unauthenticated tunnel.

```python
NGROK_AUTH_TOKEN = ""

if NGROK_AUTH_TOKEN:
    from pyngrok import ngrok
    ngrok.set_auth_token(NGROK_AUTH_TOKEN)
```

```python
from pyngrok import ngrok
import subprocess
import time

process = subprocess.Popen([
    "streamlit",
    "run",
    "streamlit_app.py",
    "--server.port",
    "8501",
    "--server.address",
    "0.0.0.0",
])
time.sleep(5)
public_url = ngrok.connect(8501)
print(f"LNPhub Streamlit app: {public_url}")
```
